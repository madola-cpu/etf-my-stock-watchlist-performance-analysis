# Market Data Collection

This notebook collects five years of historical market data for the selected ETFs and stocks using Yahoo Finance.


## 1. setup and imports

```python

import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as pyplot
from pathlib import Path
from datetime import datetime
from dateutil.relativedelta import relativedelta

```

## define of assets

```python

etfs = ['VNQ','VXUS','SCHD','VOO','QQQM','FXAIX']
stocks =['AMZN','AAPL','MSFT','NVDA']
tickers = etfs + stocks
```

## 3. Define Analysis Period

```python
from datetime import date


start_date = '2021-10-01'
end_date = date.today()
start_date = end_date - relativedelta(years=5)
```

## 4.Download the historical market data

```python
market_data = yf.download(
    tickers,
    start=start_date,
    auto_adjust=False,
    group_by="column"
)
```

## 6. Save Raw Data

```python
raw_data_path = Path("../data/raw")
raw_data_path.mkdir(parents=True, exist_ok=True)

market_data.to_csv(raw_data_path / "market_data.csv")

print("Raw market data saved successfully.")
```
